```python
class Solution:
    def groupAnagrams(self, strs: list[str]) -> list[list[str]]:
        seen={}
        for str in strs:
            sorted_str=''.join(sorted(str))
            if sorted_str not in seen:
                seen[sorted_str]=[str]
            else:
                seen[sorted_str].append(str)
        return list(seen.values())
```
