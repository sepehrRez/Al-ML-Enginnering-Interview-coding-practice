```python

def find_neighbours(grid: list[list[int]], current_node):
    shape=(len(grid), len(grid[0]))
    neighburs=[]
    if current_node[0] >0 and grid[current_node[0]-1][current_node[1]]!=-1:  # up
        neighburs.append((current_node[0]-1, current_node[1]))
    if current_node[0] < shape[0]-1 and grid[current_node[0]+1][current_node[1]]!=-1: #down
        neighburs.append((current_node[0]+1, current_node[1]))

    if current_node[1] >0 and grid[current_node[0]][current_node[1]-1]!=-1:  #left
        neighburs.append((current_node[0], current_node[1]-1))
    if current_node[1] < shape[1]-1 and grid[current_node[0]][current_node[1]+1]!=-1: #right
        neighburs.append((current_node[0], current_node[1]+1))
    return neighburs

```

```python
def heuristic(current, goal):
    return abs(goal[0]- current[0]) + abs(goal[1]-current[1])
```

```python
heuristic((0, 0), (1, 1))
```

```python
def a_star(Ingrid):
    #initial values
    grid= Ingrid["grid"]
    start=Ingrid["start"] 
    goal=Ingrid["goal"]
    open={}
    close={}
    result={}
    h= heuristic(start, goal)
    open[start]={
        "path":[start], 
        "g":0, 
        "h":h,
        "f": h 
    }
    while open:
        current_dim = min(open, key=lambda node: open[node]["f"])
        
        if current_dim == goal:
            result=open[current_dim]
            break
        
        current=open[current_dim] 
        close[current_dim]=open.pop(current_dim)
        neighbours=find_neighbours(grid=grid, current_node=current_dim)
        
        for neighbour in neighbours:
            new_g = current["g"] + grid[neighbour[0]][neighbour[1]]
            
            if neighbour in close:
                continue
            if neighbour not in open:
                temp = {}
                temp["path"] = current["path"] + [neighbour]
                temp["g"] = new_g
                temp["h"] = heuristic(neighbour, goal)
                temp["f"] = new_g + temp["h"]

                open[neighbour] = temp

            elif new_g < open[neighbour]["g"]:
                open[neighbour]["path"] = current["path"] + [neighbour]
                open[neighbour]["g"] = new_g
                open[neighbour]["f"] = new_g + open[neighbour]["h"]
    return result
```
